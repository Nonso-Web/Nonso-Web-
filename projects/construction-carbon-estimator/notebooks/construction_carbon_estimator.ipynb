# Construction Carbon Estimator
This notebook calculates an illustrative embodied-carbon total from synthetic material quantities and example emission factors. Replace all factors with verified, appropriate sources before any real project assessment.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'data' / 'material_quantities.csv').exists():
    ROOT = ROOT / 'projects' / 'construction-carbon-estimator'
quantities = pd.read_csv(ROOT / 'data' / 'material_quantities.csv')
factors = pd.read_csv(ROOT / 'data' / 'illustrative_emission_factors.csv')
display(quantities, factors)
```

```python
df = quantities.merge(factors, on='material', suffixes=('_quantity','_factor'), validate='one_to_one')
if df.unit_quantity.ne(df.unit_factor).any():
    raise ValueError('Quantity and factor units do not match')
df['estimated_kgco2e'] = df.quantity * df.illustrative_kgco2e_per_unit
display(df[['material','quantity','unit_quantity','illustrative_kgco2e_per_unit','estimated_kgco2e']])
print('Illustrative total:', round(df.estimated_kgco2e.sum(), 1), 'kgCO2e')
```

```python
df.set_index('material').estimated_kgco2e.sort_values().plot(kind='barh', figsize=(8,4), xlabel='Illustrative emissions (kgCO2e)', title='Material contribution')
plt.tight_layout()
```

## Limitations and responsible use
All values are synthetic or illustrative. This is not a whole-life carbon assessment. A real assessment requires verified EPD or recognised database factors, consistent system boundaries and units, lifecycle modules, geography, transport, waste assumptions and professional review.
