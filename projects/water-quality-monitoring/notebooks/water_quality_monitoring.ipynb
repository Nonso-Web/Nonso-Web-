# Water Quality Monitoring and Screening
Explore synthetic sample data. The screening thresholds are illustrative only and must not be interpreted as regulatory compliance limits.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / 'data' / 'water_quality_samples.csv').exists():
    ROOT = ROOT / 'projects' / 'water-quality-monitoring'
df = pd.read_csv(ROOT / 'data' / 'water_quality_samples.csv')
df.head()
```

```python
metrics = ['ph','turbidity_ntu','dissolved_oxygen_mg_l','nitrate_mg_l']
df[metrics].describe()
```

```python
# Demonstration thresholds only, not regulatory criteria.
df['ph_flag'] = ~df.ph.between(6.5,8.5)
df['turbidity_flag'] = df.turbidity_ntu > 3.0
df['nitrate_flag'] = df.nitrate_mg_l > 5.0
df[['sample_id','site','ph_flag','turbidity_flag','nitrate_flag']]
```

```python
df.groupby('site').turbidity_ntu.mean().plot(kind='bar', ylabel='Mean turbidity (NTU)', title='Synthetic sample turbidity by site');
plt.tight_layout()
```

## Limitations
All rows are synthetic. Thresholds are placeholders for demonstrating data workflows, not statutory limits. A real assessment requires verified sampling, laboratory quality assurance, appropriate standards and expert interpretation.
