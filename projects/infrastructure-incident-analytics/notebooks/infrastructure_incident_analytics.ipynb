# Infrastructure Incident Analytics
Exploratory analysis of a generated incident register. All records and coordinate positions are synthetic; the coordinates are not GPS or a map.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
data_path = Path('../data/infrastructure_incidents.csv')
df = pd.read_csv(data_path, parse_dates=['date_reported'])
df.head()
```

## Data checks and overview

```python
print('Rows:', len(df))
print('Duplicate IDs:', df['incident_id'].duplicated().sum())
print(df.isna().sum())
df.describe(include='all')
```

## Incidents by asset type

```python
asset_counts = df['asset_type'].value_counts()
asset_counts.sort_values().plot(kind='barh', figsize=(8,4), title='Synthetic incidents by asset type')
plt.xlabel('Incident count')
plt.tight_layout()
plt.show()
```

## Monthly trend

```python
monthly = df.set_index('date_reported').resample('MS').size()
monthly.plot(figsize=(9,4), title='Monthly synthetic incident counts')
plt.ylabel('Incidents')
plt.tight_layout()
plt.show()
```

## Response and repair duration by severity

```python
severity_summary = df.groupby('severity').agg(
    incident_count=('incident_id','count'),
    median_response_hours=('response_time_hours','median'),
    median_repair_hours=('repair_time_hours','median'))
severity_summary
```

## Fictional coordinate-plane plot
These positions are invented values and must not be interpreted as real-world geography.

```python
fig, ax = plt.subplots(figsize=(7,5))
for severity, group in df.groupby('severity'):
    ax.scatter(group['coord_x'], group['coord_y'], label=severity, alpha=.75)
ax.set(xlabel='Synthetic X position', ylabel='Synthetic Y position', title='Fictional coordinate plane — not a map')
ax.legend()
plt.tight_layout()
plt.show()
```

## Interpretation
Describe patterns as properties of this generated sample only. Do not infer actual service performance, risk, causes or geographic hotspots.
